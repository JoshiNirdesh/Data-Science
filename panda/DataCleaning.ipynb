```python
import pandas as pd 
data = {
    "name": ["Ram", "Sita", "Hari", "Gita", "John", "Ram"],
    "age": [22, None, 25, 17, 30, 22],
    "salary": [30000, 25000, None, 20000, 60000, 30000],
    "city": ["Kathmandu", "Pokhara", "Kathmandu", None, "Bhaktapur", "Kathmandu"]
}

df = pd.DataFrame(data)
```

```python
df.isna()
```

```python
df.isna().sum() 
```

```python
df.dropna()
```

```python
df['age']=df['age'].fillna(df['age'].mean())
df
```

```python
df['salary']=df['salary'].fillna(0)
df
```

```python
df['city']=df['city'].fillna("Unknown")
df
```

```python
df.drop_duplicates()
```

```python
df.rename(columns={"salary":"monthly_salary"},inplace=True)
```

```python
df.dtypes
df['monthly_salary']=df['monthly_salary'].astype(int)
```

```python
df['annual_salary']=df['monthly_salary']*12
df
```
